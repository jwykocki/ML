## Exercise 1. Build your own tokenizer

Build two different tokenizers:
- ``tokenize_sentence``: function tokenizing text into sentences,
- ``tokenize_word``: function tokenizing text into words.

```python
from typing import List

def tokenize_words(text: str) -> list:
    pattern = r"\b[\w']+\b"
    return re.findall(pattern, text.lower())

def tokenize_sentence(text: str) -> list:
    pattern = r'(?<=[.!?])\s+(?=[A-Z])'
    sentences = re.split(pattern, text.strip())
    return [s.strip() for s in sentences]

text = "Here we go again. I was supposed to add this text later.\
Well, it's 10.p.m. here, and I'm actually having fun making this course. :o\
I hope you are getting along fine with this presentation, I really did try.\
And one last sentence, just so you can test you tokenizers better."

print("Tokenized sentences:")
print(tokenize_sentence(text))

print("Tokenized words:")
print(tokenize_words(text))
```

## Exercise 2. Get tags from Trump speech using NLTK

You should use the ``trump.txt`` file, read it and find the tags for each word. Use NLTK for it.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

file = open("../datasets/trump.txt", "r", encoding="utf-8")
trump = file.read()
words = word_tokenize(trump)

tags = pos_tag(words)

print("\nFirst 20 words with tags:")
for word, tag in tags[:20]:
    print(f"{word}: {tag}")

unique_tags = set(tag for _, tag in tags)
print(f"\nNumber of unique tags: {len(unique_tags)}")
print(f"Unique tags: {unique_tags}")
```

## Exercise 4. Build your own Bag Of Words implementation using tokenizer created before 

You need to implement following methods:

- ``fit_transform`` - gets a list of strings and returns matrix with it's BoW representation
- ``get_features_names`` - returns list of words corresponding to columns in BoW

```python
import numpy as np
import spacy
from collections import Counter

class BagOfWords:

    def __init__(self):
        self.__nlp = spacy.load("en_core_web_sm")
        self.__bow_list = []
        self.__vocabulary = {}

    def fit_transform(self, corpus: list):

        tokenized_docs = []

        for doc_text in corpus:
            doc = self.__nlp(doc_text.lower())
            tokens = [token.lemma_ for token in doc if not token.is_punct and not token.is_space]
            tokenized_docs.append(tokens)

            for token in tokens:
                if token not in self.__vocabulary:
                    self.__vocabulary[token] = len(self.__vocabulary)

        n_docs = len(corpus)
        n_features = len(self.__vocabulary)
        bow_matrix = np.zeros((n_docs, n_features))

        for i, tokens in enumerate(tokenized_docs):
            token_counts = Counter(tokens)
            for token, count in token_counts.items():
                if token in self.__vocabulary:
                    bow_matrix[i, self.__vocabulary[token]] = count

        self.__bow_list = tokenized_docs
        return bow_matrix

    def get_feature_names(self) -> list:

        sorted_vocab = sorted(self.__vocabulary.items(), key=lambda x: x[1])
        return [word for word, _ in sorted_vocab]

corpus = [
     'Bag Of Words is based on counting',
     'words occurences throughout multiple documents.',
     'This is the third document.',
     'As you can see most of the words occur only once.',
     'This gives us a pretty sparse matrix, see below. Really, see below',
]

vectorizer = BagOfWords()
X = vectorizer.fit_transform(corpus)
print("BoW Matrix shape:", X.shape)
print("BoW Matrix:\n", X)
print("\nFeature names:", vectorizer.get_feature_names())
print("Number of features:", len(vectorizer.get_feature_names()))
```

```python
from nltk.book import *
import random
import re

wall_street = text7.tokens

def cleanup_tokens():
    compiled_pattern = re.compile(r"^[a-zA-Z0-9.!?]+$")
    clean = list(filter(compiled_pattern.match, tokens))
    return clean

tokens = wall_street
N = 5
SEP = " "

def build_ngrams():
    ngrams = []
    for i in range(len(tokens)-N+1):
        ngrams.append(tokens[i:i+N])
    return ngrams

def ngram_freqs(ngrams):
    counts = {}
    for ngram in ngrams:
        token_seq = SEP.join(ngram[:-1])
        last_token = ngram[-1]

        if token_seq not in counts:
            counts[token_seq] = {}

        if last_token not in counts[token_seq]:
            counts[token_seq][last_token] = 0

        counts[token_seq][last_token] += 1

    return counts

def next_word(text, N, counts):
    token_seq = SEP.join(text.split()[-(N-1):])
    choices = counts[token_seq].items()
    total = sum(weight for choice, weight in choices)
    r = random.uniform(0, total)
    upto = 0
    for choice, weight in choices:
        upto += weight
        if upto > r:
            return choice
    assert False

def clean_generated(text):
    sentences = re.split(r'([.!?])', text)

    cleaned_parts = []
    for i in range(0, len(sentences)-1, 2):
        if i+1 < len(sentences):
            sentence = sentences[i].strip()
            punctuation = sentences[i+1]

            sentence = sentence.rstrip()

            if sentence:
                sentence = sentence[0].upper() + sentence[1:] if len(sentence) > 1 else sentence.upper()

            cleaned_parts.append(sentence + punctuation)

    if len(sentences) % 2 == 1 and sentences[-1]:
        cleaned_parts.append(sentences[-1])

    return ' '.join(cleaned_parts)

ngrams = build_ngrams()
print(f"Number of 5-grams: {len(ngrams)}")

counts = ngram_freqs(ngrams)
start_seq = "We have"
print(f"Available starting sequences: {list(counts.keys())[:5]}...")

sentence_count = 5
generated = start_seq.lower()
sentences = 0

while sentences < sentence_count and generated.split()[-1] in counts.get(SEP.join(generated.split()[-(N-1):]), {}):
    try:
        next_word_text = next_word(generated, N, counts)
        generated += SEP + next_word_text
        sentences += 1 if generated.endswith(('.', '!', '?')) else 0
    except (KeyError, AssertionError):
        break

cleaned_text = clean_generated(generated)
print(cleaned_text)
print("\nOriginal text length:", len(generated))
print("Cleaned text length:", len(cleaned_text))

if len(generated) > 200:
    print("Before cleanup:", generated[:200] + "...")
    print("After cleanup: ", cleaned_text[:200] + "...")
```
